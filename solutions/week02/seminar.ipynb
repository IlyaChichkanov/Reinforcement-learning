## 2. Модель среды: таблицы $p$ и $r$

Как на лекции (раздел 2.1), вся среда — это два массива:

* `P[s, a, s_next]` — вероятность попасть в клетку `s_next`, выбрав действие `a` в клетке `s`;
* `R[s, a]` — средняя награда за такой шаг.

Собрать их можно, не запуская среду: метод `env.transitions(s, a)` возвращает список исходов, по одному кортежу `(вероятность, следующая клетка, награда, конец эпизода)` на исход. Например:

```python
>>> env.transitions(10, 3)
[(0.8, 5, -0.04, False), (0.1, 10, -0.04, False), (0.1, 11, -0.04, False)]
```

Пройдите по всем клеткам и действиям и разложите эти исходы по двум массивам. Помните, что в `R[s, a]` попадает **средняя** награда: каждую награду нужно домножить на вероятность её исхода.

````python
def mdp_matrices(env):
    """Модель среды: P[s, a, s_next] — вероятности переходов, R[s, a] — средняя награда за шаг."""
    P = np.zeros((env.n_states, env.n_actions, env.n_states))
    R = np.zeros((env.n_states, env.n_actions))
    for s in range(env.n_states):
        for a in range(env.n_actions):
            for prob, s_next, reward, _ in env.transitions(s, a):
                P[s, a, s_next] += prob
                R[s, a] += prob * reward
    return P, R

# Проверка 2
P, R = mdp_matrices(env)
assert np.allclose(P.sum(axis=2), 1.0), "из каждой пары (клетка, действие) вероятности должны суммироваться в 1"
P_ref, R_ref = mdp_matrices_reference(env)
assert np.allclose(P, P_ref) and np.allclose(R, R_ref), "таблицы не совпали с эталоном"
print(f"OK: P имеет форму {P.shape}, R — {R.shape}")
print(f"P[{env.start}, ↑] ненулевые:", {int(s): round(float(P[env.start, 3, s]), 2) for s in np.flatnonzero(P[env.start, 3])})
````

## 3. Одна итерация уравнения Беллмана

Формула из раздела 4.1 лекции:

$$
V_k(s') = \max_{a'} Q_k(s', a'), \qquad
Q_{k+1}(s, a) = r(s, a) + \gamma \sum_{s'} p(s' \mid s, a)\, V_k(s').
$$

Напишите её в векторном виде — это две строки. Проверка сравнит ваш результат с той же формулой, расписанной явными циклами.

````python
def bellman_update(Q, P, R, gamma=GAMMA):
    """Одна итерация: подставляем приближение Q в правую часть уравнения оптимальности."""
    V = Q.max(axis=1)
    return R + gamma * (P @ V)

def bellman_update_loops(Q, P, R, gamma=GAMMA):
    """Та же формула, расписанная циклами — дана для проверки."""
    V, Q_new = Q.max(axis=1), np.zeros_like(Q)
    for s in range(P.shape[0]):
        for a in range(P.shape[1]):
            Q_new[s, a] = R[s, a] + gamma * sum(P[s, a, s_next] * V[s_next] for s_next in range(P.shape[2]))
    return Q_new

# Проверка 3
Q_zero = np.zeros((env.n_states, env.n_actions))
for Q_test in (Q_zero, np.random.default_rng(0).normal(size=Q_zero.shape)):
    assert np.allclose(bellman_update(Q_test, P, R), bellman_update_loops(Q_test, P, R)), "ваша итерация не совпала с формулой"
print("OK: одна итерация из нулей, клетка рядом с выходом:", bellman_update(Q_zero, P, R).max(axis=1).round(2)[:5])
````

## 4. Value iteration

Теперь повторяем итерацию, пока приближение не перестанет меняться (раздел 4.2 лекции). Условие остановки: наибольшая поправка меньше `tol`.

Функция нигде не смотрит на сам лабиринт — только на `P` и `R`. Поэтому в домашке (задача 4.2) она без изменений решит задачу склада.

````python
def value_iteration(P, R, gamma=GAMMA, tol=1e-10, max_iter=1000):
    """Решение уравнения оптимальности. Возвращает Q* и число сделанных итераций."""
    Q = np.zeros_like(R)
    for k in range(1, max_iter + 1):
        Q_new = bellman_update(Q, P, R, gamma)
        if np.abs(Q_new - Q).max() < tol:
            return Q_new, k
        Q = Q_new
    return Q, max_iter

# Проверка 4
Q_star, n_iter = value_iteration(P, R)
V_star = Q_star.max(axis=1)
pi_star = greedy_policy(Q_star)
assert np.allclose(Q_star, solve_q_star(P, R), atol=1e-6), "решение не совпало с эталоном"
assert average_return(env, pi_star, np.random.default_rng(0), 300) > average_return(env, uniform_policy(env), np.random.default_rng(0), 300)
print(f"OK: сошлось за {n_iter} итераций; ценность старта {V_star[env.start]:+.2f}")
print(f"средний return: жадная стратегия {average_return(env, pi_star, np.random.default_rng(0), 300):+.2f}, "
      f"случайная {average_return(env, uniform_policy(env), np.random.default_rng(0), 300):+.2f}")

fig, axes = plt.subplots(1, 2, figsize=(9, 3.2))
plot_q(Q_star, env, ax=axes[0], title="Q*(s, a)")
draw_policy(pi_star, env, ax=axes[1], values=V_star, title="V* и оптимальная стратегия")
plt.show()
````

## 5. Оценка заданной стратегии

### 5.1 Итерациями

До сих пор мы искали **лучшую** стратегию. Теперь оценим **любую** — для этого в уравнении вместо максимума стоит усреднение по стратегии (разделы 3.1 и 4 лекции):

$$
V^\pi(s') = \sum_{a'} \pi(a' \mid s')\, Q^\pi(s', a'), \qquad
Q^\pi(s, a) = r(s, a) + \gamma \sum_{s'} p(s' \mid s, a)\, V^\pi(s').
$$

Отличие от раздела 3 — одна строка: вместо `Q.max(axis=1)` берём `(policy * Q).sum(axis=1)`.

````python
def evaluate_policy(policy, P, R, gamma=GAMMA, n_iter=300):
    """Ценность заданной стратегии. Возвращает Q^π и V^π."""
    Q = np.zeros_like(R)
    for _ in range(n_iter):
        V = (policy * Q).sum(axis=1)
        Q = R + gamma * (P @ V)
    return Q, (policy * Q).sum(axis=1)

# Проверка 5: ценность случайной стратегии, посчитанная уравнением, совпадает с Монте-Карло
uniform = uniform_policy(env)
Q_uniform, V_uniform = evaluate_policy(uniform, P, R)

rng = np.random.default_rng(1)
V_mc = estimate_values_mc([run_episode(env, uniform, rng) for _ in range(3000)], env.n_states)
visited = ~np.isnan(V_mc)
assert abs(V_uniform[env.start] - V_mc[env.start]) < 0.05, "ценность старта разошлась с Монте-Карло"
assert np.abs(V_uniform[visited] - V_mc[visited]).max() < 0.15, "оценка сильно разошлась с Монте-Карло"
print(f"OK: наибольшее расхождение с Монте-Карло {np.abs(V_uniform[visited] - V_mc[visited]).max():.3f}")

fig, axes = plt.subplots(1, 2, figsize=(9, 3.2))
plot_values(V_mc, env, ax=axes[0], title="3000 эпизодов Монте-Карло")
plot_values(V_uniform, env, ax=axes[1], title="решение уравнения")
plt.show()
````

### 5.2 Точно: одна линейная система

Для заданной стратегии в уравнении нет максимума, поэтому оно **линейно** (раздел 7 лекции). Усредним модель по стратегии:

$$
P_\pi[s, s'] = \sum_a \pi(a \mid s)\, p(s' \mid s, a), \qquad r_\pi[s] = \sum_a \pi(a \mid s)\, r(s, a),
$$

тогда $V^\pi = r_\pi + \gamma P_\pi V^\pi$, то есть $(I - \gamma P_\pi)\, V^\pi = r_\pi$ — система из `n_states` уравнений, которую `np.linalg.solve` решает за одну строку, без итераций.

**TODO 5.2.** Напишите `evaluate_policy_exact(policy, P, R, gamma)`, возвращающую $V^\pi$. Подсказка: $P_\pi$ удобно собрать через `np.einsum("sa,sat->st", policy, P)`. В домашке (задача 4.1) эта функция посчитает ценность ручной политики склада.

````python
def evaluate_policy_exact(policy, P, R, gamma=GAMMA):
    """V^π как решение линейной системы (I - γ P_π) V = r_π."""
    P_pi = np.einsum("sa,sat->st", policy, P)
    r_pi = (policy * R).sum(axis=1)
    return np.linalg.solve(np.eye(len(r_pi)) - gamma * P_pi, r_pi)

# Проверка 5.2: точное решение совпадает с итерационным
V_exact = evaluate_policy_exact(uniform, P, R)
assert V_exact.shape == (env.n_states,)
assert np.allclose(V_exact, V_uniform, atol=1e-6), "точная ценность разошлась с итерационной"
assert np.allclose(evaluate_policy_exact(greedy_policy(Q_star), P, R), V_star, atol=1e-6), "для оптимальной стратегии должно получиться V*"
print(f"OK: ценность старта случайной стратегии {V_exact[env.start]:+.3f}, оптимальной {V_star[env.start]:+.3f}")
````

## 6. Улучшение стратегии

### 6.1 Один шаг

Имея $Q^\pi$, построим новую стратегию: в каждой клетке выбираем действие с наибольшим $Q^\pi$. Проверка убедится, что ценность не упала **ни в одной** клетке — это и есть гарантия шага улучшения из лекции.

````python
def improve(policy, P, R, gamma=GAMMA):
    """Один шаг улучшения: оценить стратегию и пересесть на жадные действия по Q^π."""
    Q_pi, _ = evaluate_policy(policy, P, R, gamma)
    return greedy_policy(Q_pi)

# Проверка 6
improved = improve(uniform, P, R)
Q_improved, V_improved = evaluate_policy(improved, P, R)
assert np.all(V_improved >= V_uniform - 1e-9), "после шага улучшения ценность не должна падать ни в одной клетке"
assert average_return(env, improved, np.random.default_rng(2), 300) > average_return(env, uniform, np.random.default_rng(2), 300)
print(f"OK: ценность старта {V_uniform[env.start]:+.2f} → {V_improved[env.start]:+.2f} "
      f"(оптимум {V_star[env.start]:+.2f})")

fig, axes = plt.subplots(1, 2, figsize=(9, 3.2))
draw_policy(uniform, env, ax=axes[0], values=V_uniform, title="было: случайная стратегия")
draw_policy(improved, env, ax=axes[1], values=V_improved, title="стало: жадная по Q^π")
plt.show()
````

### 6.2 ★ Policy iteration

Сколько шагов улучшения нужно на вашем лабиринте, чтобы дойти до оптимума? Повторяйте «оценить — улучшить», пока стратегия не перестанет меняться: это и есть **policy iteration**, подробно — на неделе 3.

**TODO 6.2 (★, по желанию).** `policy_iteration(P, R)` начинает со случайной стратегии, применяет `improve`, пока новая стратегия отличается от старой, и возвращает итоговую стратегию и число сделанных шагов.

````python
def policy_iteration(P, R, gamma=GAMMA, max_iter=100):
    """Повторять improve, пока стратегия меняется. Возвращает стратегию и число шагов."""
    policy = uniform_policy(env)
    for k in range(1, max_iter + 1):
        new_policy = improve(policy, P, R, gamma)
        if np.array_equal(new_policy, policy):
            return policy, k
        policy = new_policy
    return policy, max_iter

# Проверка 6.2
pi_pi, n_steps = policy_iteration(P, R)
assert np.allclose(evaluate_policy_exact(pi_pi, P, R), V_star, atol=1e-6), "policy iteration должна прийти к оптимальной ценности"
print(f"OK: policy iteration сошлась, шагов улучшения: {n_steps}; у value iteration итераций: {n_iter}")
````

## 7. Лабиринт как среда Gymnasium

`GridWorld` из лекции похож на среду Gymnasium, но ею не является: у него нет пространств `observation_space` и `action_space`, а случайность живёт в собственном генераторе. Оформим лабиринт как настоящий `gym.Env` — ровно так в домашке (задача 3.1) вы напишете склад.

**TODO 7.1.** Допишите `MazeEnv`:

* `observation_space` и `action_space` — `spaces.Discrete` нужного размера;
* `reset(seed)` вызывает `super().reset(seed=seed)` (это заводит генератор `self.np_random`), ставит агента на старт, обнуляет счётчик шагов и возвращает `(наблюдение, info)`;
* `step(action)` берёт исходы `self.grid.transitions(s, a)`, выбирает один с их вероятностями через `self.np_random.choice` и возвращает `(наблюдение, награда, terminated, truncated, info)`. `terminated` — эпизод кончился по правилам (выход или яма), `truncated` — упёрлись в лимит `max_steps`.

Среда не хранит свою копию правил: она **разыгрывает** ту же модель, что лежит в `P` и `R`. Последняя проверка это и подтверждает — ценность оптимальной стратегии, измеренная в новой среде по Монте-Карло, совпадает с $V^*$.

````python
class MazeEnv(gym.Env):
    """Лабиринт из лекции в интерфейсе Gymnasium."""

    def __init__(self, grid, max_steps=100):
        self.grid, self.max_steps = grid, max_steps
        self.n_states, self.n_actions = grid.n_states, grid.n_actions   # чтобы работали run_episode и average_return
        self.observation_space = spaces.Discrete(grid.n_states)
        self.action_space = spaces.Discrete(grid.n_actions)

    def reset(self, seed=None, options=None):
        super().reset(seed=seed)
        self.s, self.t = self.grid.start, 0
        return self.s, {}

    def step(self, action):
        outcomes = self.grid.transitions(self.s, int(action))
        k = self.np_random.choice(len(outcomes), p=[prob for prob, *_ in outcomes])
        _, self.s, reward, terminated = outcomes[k]
        self.t += 1
        truncated = self.t >= self.max_steps and not terminated
        return self.s, float(reward), terminated, truncated, {}


# Проверка 7.1: интерфейс, воспроизводимость, частоты переходов и ценность
maze = MazeEnv(env)
check_env(maze, skip_render_check=True)

def first_steps(seed, n=200, action=3):
    """Куда попадаем за один шаг из старта, n раз подряд; сид задаётся один раз."""
    e = MazeEnv(env); e.reset(seed=seed)
    out = []
    for _ in range(n):
        e.reset(); out.append(e.step(action)[0])
    return out

assert first_steps(0) == first_steps(0) and first_steps(0) != first_steps(1), "среда должна быть воспроизводима по сиду"
freq = np.bincount(first_steps(0, n=3000), minlength=env.n_states) / 3000
assert np.abs(freq - P[env.start, 3]).max() < 0.03, "частоты исходов из старта должны совпасть с P[start, ↑]"
mc = average_return(maze, pi_star, np.random.default_rng(0), 2000)
assert abs(mc - V_star[env.start]) < 0.05, f"Монте-Карло в MazeEnv даёт {mc:.3f}, а V*(старт) = {V_star[env.start]:.3f}"
print(f"OK: частоты из старта {dict((int(s), round(float(f), 2)) for s, f in enumerate(freq) if f > 0)}, "
      f"P[start, ↑] = {dict((int(s), round(float(p), 2)) for s, p in enumerate(P[env.start, 3]) if p > 0)}")
print(f"    ценность старта: Монте-Карло {mc:+.3f}, V* {V_star[env.start]:+.3f}")
````
